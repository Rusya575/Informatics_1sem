```python
def find_previous(arr):
    result = []
    for c in range(len(arr)):
        prev= -1
        for j in range(c):
            if arr[j] >= arr[c]:
                prev = j
        result.append(prev)
    print(' '.join(map(str, result)))
input_array = list(map(int, input().split()))
find_previous(input_array)
```

```python
sky_high = list(map(int, input().split()))
ans = []

for i in range(len(sky_high)):
    prev_sky = -1
    for j in range(i):
        if sky_high[j] >= sky_high[i]:
            prev_sky = j
    ans.append(prev_sky)
print(*ans)
```

```python
def count_friends(arr):
    result = ''
    length = len(arr)
    for i in range(length):
        count = 0
        m = arr[i]
        for j in range(i, length):
            if arr[j] > m:
                count += 1
                m = arr[j]
        result += f"{count} "
    print(result.strip())

input_list = list(map(int, input().split()))
count_friends(input_list)
```

```python
pas_num = list(map(int, input().split()))
ans = ''

for i in range(len(pas_num)):
    chain = 0
    max_num = pas_num[i]
    for j in range(i, len(pas_num)):
        if pas_num[j] > max_num:
            chain += 1
            max_num = pas_num[j]
    ans += f"{chain}"
    
print(ans.strip())
```
