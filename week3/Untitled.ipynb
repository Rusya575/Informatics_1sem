```python
#упражнение 1

n = int(input())

def fib(n):
    if n == 0:
        return 1
    elif n == 1:
        return 1
    else:
        return fib(n-2) + fib(n-1)
print(fib(n))
```

```python
#упражнение 2

def sf(n, a = []):
    for i in range(2, n):
        if n % i == 0:
            a.append(i)
            return sf(n//i, a)
    a.append(n)
    return a

print(*sf(int(input())))
```

```python
#упраднение 5

import numpy as np

def matrix(m, n):
    a = np.zeros((m, n))
    a[0][0] = 1
    i = 0
    j = 1
    counter = 1

    while 0 in a:
        while j < n and a[i][j] == 0:
            a[i][j] = counter + 1
            counter += 1
            j += 1
        j -= 1
        i += 1

        while i < m and a[i][j] == 0:
            a[i][j] = counter + 1
            counter += 1
            i += 1
        i -= 1
        j -= 1

        while j > -1 and a[i][j] == 0:
            a[i][j] = counter + 1
            counter += 1
            j -= 1
        j += 1
        i -= 1

        while i > -1 and a[i][j] == 0:
            a[i][j] = counter + 1
            counter += 1
            i -= 1
        i += 1
        j += 1
    return a

def multiplication(first_res):
    for i in range(len(first_res)):
        first_res[i] = np.array(first_res[i]) * (i + 1)
    return first_res

data = list(map(int, input().split()))
line = data[0]
pillar = data[1]

mat = matrix(line, pillar)

print(multiplication(mat))
```
