```python
n = list(map(str, input().split()))
s = n[1]
n = int(n[0])
l = len(s)
k = l//n
list_1 = []
z = 0

for i in range(k):
    list_1.append(s[z:(z+n)])
    list_1[i] = (list_1[i][::-1])
    z += n
print(''.join(list_1))
```

```python
a = input()
b = a
alph = "AHIMOTUVWXY18EJSZ3L25"
l = list(a)
fl = True

for i in range(len(l)):
    if l[i] not in alph:
        fl = False

if "E" in a:
    a = a.replace("E", "3")
if "J" in a:
    a = a.replace("J", "L")
if "S" in a:
    a = a.replace("S", "2")
if "Z" in a:
    a = a.replace("Z", "5")

if a == a[::-1] and fl:
    print(b, "is a mirrored palindrome.")
elif a == a[::-1]:
    print(b, "is a regular palindrome.")
elif a == a[::-1]:
    print(b, "is a mirrored string.")
else:
    print(b, "is not a palindrome.")
```
